```python
import pandas as pd
import numpy as np
df = pd.read_csv("Student Social Media And Mental Health Impact.csv")
df
# Count missing values in each column
df.isna().sum()

# Display rows containing missing values
df[df.isna().any(axis=1)]
df['Avg_Daily_Usage_Hours'] = df['Avg_Daily_Usage_Hours'].fillna(df['Avg_Daily_Usage_Hours'].mean())

df['Study_Hours'] = df['Study_Hours'].fillna(df['Study_Hours'].median())
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

df['Gender'] = df['Gender'].fillna("Unknown")
# Remove rows with any missing values
df.dropna(inplace=True)

# OR remove columns with missing values
df.dropna(axis=1, inplace=True)
df
```

```python
import pandas as pd

# Assume the dataset is loaded into a DataFrame named df

# =====================================================
# 1. BOOLEAN INDEXING
# =====================================================

# Age greater than 20
df[df['Age'] > 20]

# High stress users
df[df['Stress_Level'] == 'High']

# Sleep less than 6 hours
df[df['Sleep_Hours_Per_Night'] < 6]

# Multiple conditions
df[(df['Country'] == 'India') & (df['Study_Hours'] > 4)]


# =====================================================
# 2. QUERY METHOD
# =====================================================

# Average daily usage greater than 5 hours
df.query("Avg_Daily_Usage_Hours > 5")

# Mental health score greater than 7
df.query("Mental_Health_Score > 7")

# Multiple conditions
df.query("Country == 'India' and Study_Hours > 4")


# =====================================================
# 3. isin()
# =====================================================

# Country is India or Canada
df[df['Country'].isin(['India', 'Canada'])]

# Platform is Instagram or Facebook
df[df['Most_Used_Platform'].isin(['Instagram', 'Facebook'])]


# =====================================================
# 4. between()
# =====================================================

# Study hours between 3 and 6
df[df['Study_Hours'].between(3, 6)]

# Sleep hours between 6 and 8
df[df['Sleep_Hours_Per_Night'].between(6, 8)]

# Mental health score between 5 and 8
df[df['Mental_Health_Score'].between(5, 8)]


# =====================================================
# 5. loc (Label-Based Selection)
# =====================================================

# Mental health score greater than 7
df.loc[df['Mental_Health_Score'] > 7]

# Select specific columns
df.loc[df['Country'] == 'India',
       ['Age', 'Gender', 'Study_Hours']]

# Multiple conditions
df.loc[
    (df['Stress_Level'] == 'Low') &
    (df['Study_Hours'] > 5)
]


# =====================================================
# 6. iloc (Position-Based Selection)
# =====================================================

# First 5 rows
df.iloc[:5]

# Rows 10–20 and first 4 columns
df.iloc[10:21, 0:4]

# Single row
df.iloc[15]

# Single value (3rd row, 4th column)
df.iloc[2, 3]


# =====================================================
# COMBINED FILTERING EXAMPLES
# =====================================================

# Indian students using Instagram
df[
    (df['Country'] == 'India') &
    (df['Most_Used_Platform'] == 'Instagram')
]

# High stress and low sleep
df[
    (df['Stress_Level'] == 'High') &
    (df['Sleep_Hours_Per_Night'] < 6)
]

# Study more than 5 hours and social media usage less than 3 hours
df[
    (df['Study_Hours'] > 5) &
    (df['Avg_Daily_Usage_Hours'] < 3)
]

# Undergraduate students from Canada
df[
    (df['Academic_Level'] == 'Undergraduate') &
    (df['Country'] == 'Canada')
]

# Good physical activity and low stress
df[
    (df['Physical_Activity_Hours'] == 'good') &
    (df['Stress_Level'] == 'Low')
]

# Female users with mental health score above 8
df[
    (df['Gender'] == 'Female') &
    (df['Mental_Health_Score'] > 8)
]
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
# df = pd.read_csv("social_media_dataset.csv")

# -------------------------------
# 1. Line Plot
# -------------------------------
plt.figure(figsize=(8,5))
plt.plot(df.index[:100], df['Avg_Daily_Usage_Hours'][:100],
         color='pink', marker='o', linewidth=2)
plt.title("Average Daily Usage Hours")
plt.xlabel("Student Index")
plt.ylabel("Usage Hours")
plt.grid(True)
plt.show()


# -------------------------------
# 2. Bar Chart
# -------------------------------
platform_counts = df['Most_Used_Platform'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(platform_counts.index, platform_counts.values, color='skyblue')
plt.title("Most Used Social Media Platforms")
plt.xlabel("Platform")
plt.ylabel("Number of Users")
plt.xticks(rotation=45)
plt.show()


# -------------------------------
# 3. Histogram
# -------------------------------
plt.figure(figsize=(8,5))
plt.hist(df['Study_Hours'], bins=10,
         color='green', edgecolor='black')
plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.show()


# -------------------------------
# 4. Scatter Plot
# -------------------------------
plt.figure(figsize=(8,5))
plt.scatter(df['Avg_Daily_Usage_Hours'],
            df['Mental_Health_Score'],
            color='red', alpha=0.6)

plt.title("Social Media Usage vs Mental Health")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Mental Health Score")
plt.show()


# -------------------------------
# 5. Pie Chart
# -------------------------------
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_counts.values,
        labels=gender_counts.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Gender Distribution")
plt.show()


# -------------------------------
# 6. Box Plot
# -------------------------------
plt.figure(figsize=(6,5))
plt.boxplot(df['Sleep_Hours_Per_Night'])
plt.title("Sleep Hours Box Plot")
plt.ylabel("Hours")
plt.show()


# -------------------------------
# 7. Multiple Line Plot
# -------------------------------
plt.figure(figsize=(8,5))
plt.plot(df.index[:100], df['Study_Hours'][:100],
         label='Study Hours')
plt.plot(df.index[:100], df['Sleep_Hours_Per_Night'][:100],
         label='Sleep Hours')

plt.title("Study Hours vs Sleep Hours")
plt.xlabel("Student Index")
plt.ylabel("Hours")
plt.legend()
plt.grid(True)
plt.show()


# -------------------------------
# 8. Horizontal Bar Chart
# -------------------------------
stress = df['Stress_Level'].value_counts()

plt.figure(figsize=(7,5))
plt.barh(stress.index, stress.values, color='orange')
plt.title("Stress Level Distribution")
plt.xlabel("Count")
plt.show()


# -------------------------------
# 9. Area Plot
# -------------------------------
plt.figure(figsize=(8,5))
plt.fill_between(df.index[:100],
                 df['Avg_Daily_Usage_Hours'][:100],
                 color='lightblue')

plt.title("Area Plot of Daily Usage")
plt.xlabel("Student Index")
plt.ylabel("Usage Hours")
plt.show()


# -------------------------------
# 10. Subplots
# -------------------------------
fig, ax = plt.subplots(1, 2, figsize=(12,5))

ax[0].hist(df['Study_Hours'],
           color='purple',
           edgecolor='black')
ax[0].set_title("Study Hours")

ax[1].hist(df['Sleep_Hours_Per_Night'],
           color='green',
           edgecolor='black')
ax[1].set_title("Sleep Hours")

plt.tight_layout()
plt.show()
```
