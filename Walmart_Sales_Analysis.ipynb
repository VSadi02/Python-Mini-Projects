```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("Walmart_Sales.csv")

# Display first 5 rows
print(data.head())
```

```python
# Convert Date column
data["Date"] = pd.to_datetime(
    data["Date"],
    format="%d-%m-%Y"
)

# LINE CHART
sales_trend = data.groupby("Date")["Weekly_Sales"].sum()

plt.figure(figsize=(10, 5))
plt.plot(sales_trend.index, sales_trend.values)

plt.title("Walmart Weekly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ---------------- BAR CHART ----------------

store_sales = data.groupby("Store")["Weekly_Sales"].sum().head(10)

plt.bar(store_sales.index, store_sales.values)

plt.title("Top 10 Store Sales")
plt.xlabel("Store")
plt.ylabel("Weekly Sales")

plt.show()
```

```python

# ---------------- SCATTER PLOT ----------------

plt.scatter(
    data["Temperature"],
    data["Weekly_Sales"]
)

plt.title("Temperature vs Sales")
plt.xlabel("Temperature")
plt.ylabel("Weekly Sales")

plt.show()
```

```python
# ---------------- HISTOGRAM ----------------

plt.hist(
    data["Weekly_Sales"],
    bins=10
)

plt.title("Weekly Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
# ---------------- HEATMAP ----------------

corr = data[
    [
        "Weekly_Sales",
        "Temperature",
        "Fuel_Price",
        "CPI",
        "Unemployment"
    ]
].corr()


sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Walmart Sales Correlation")

plt.show()
```

```python

# ---------------- VIOLIN PLOT ----------------

sns.violinplot(
    x="Holiday_Flag",
    y="Weekly_Sales",
    data=data
)

plt.title("Holiday Impact on Sales")

plt.show()
```

```python

# ---------------- PIE CHART ----------------

store = data.groupby("Store")["Weekly_Sales"].sum().head(5)


plt.pie(
    store,
    labels=store.index,
    autopct="%1.1f%%"
)

plt.title("Top 5 Stores Sales Contribution")

plt.show()
```
